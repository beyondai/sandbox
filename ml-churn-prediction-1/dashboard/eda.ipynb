# EDA: Churn Prediction

Real executed profiling of `../monkey-mode/data.csv`.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../monkey-mode/data.csv')
print('shape:', df.shape)
df.head()
```

## Nulls

```python
df.isna().mean().to_frame('null_rate')
```

## Target balance (`churned`)

```python
df['churned'].value_counts(normalize=True)
```

## Numeric distributions

```python
numeric_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c]) and c != 'churned']
df[numeric_cols].describe().T
```

## Categorical: plan_tier

```python
df['plan_tier'].value_counts(normalize=True)
```

## Skew check

```python
df[numeric_cols].skew().sort_values(key=abs, ascending=False)
```
