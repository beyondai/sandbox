# EDA — Kaggle Churn

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../modeling/datasets/churn_train.csv')
df.shape
```

## Nulls

```python
(df.isnull().mean() * 100).round(4)
```

## Target balance

```python
df['Churn'].value_counts(normalize=True).round(4)
```

## Numeric distributions

```python
numeric_cols = ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges']
df[numeric_cols].describe()
```

## Categorical cardinality

```python
categorical_cols = [c for c in df.select_dtypes(include='object').columns if c != 'Churn']
{c: df[c].nunique() for c in categorical_cols}
```

## TotalCharges vs. tenure * MonthlyCharges collinearity

```python
np.corrcoef(df['TotalCharges'], df['tenure'] * df['MonthlyCharges'])[0, 1]
```

## Contract and PaymentMethod breakdown

```python
df['Contract'].value_counts(), df['PaymentMethod'].value_counts()
```
